```python

import type { Product, PriceResult } from "./types";

export function calculatePrice(product: Product): PriceResult {
  const availableCompetitors = product.competitors.filter(
    (competitor) =>
      competitor.inStock &&
      Number.isFinite(competitor.price) &&
      competitor.price >= 0
  );

  // Без конкурентов сохраняем свою цену в допустимом диапазоне.
  if (availableCompetitors.length === 0) {
    return {
      sku: product.sku,
      oldPrice: product.myPrice,
      newPrice: Math.round(
        Math.min(product.maxPrice, Math.max(product.minPrice, product.myPrice))
      ),
      reason: "NO_COMPETITORS",
    };
  }

  // Находим самую низкую цену среди конкурентов в наличии.
  const cheapestCompetitor = Math.min(
    ...availableCompetitors.map((competitor) => competitor.price)
  );

  // Округляем вниз, чтобы цена была дешевле конкурента
  // как минимум на указанное значение step.
  const targetPrice = Math.floor(cheapestCompetitor - product.step);

  if (targetPrice <= product.minPrice) {
    return {
      sku: product.sku,
      oldPrice: product.myPrice,
      newPrice: product.minPrice,
      reason: "AT_MIN",
    };
  }

  if (targetPrice >= product.maxPrice) {
    return {
      sku: product.sku,
      oldPrice: product.myPrice,
      newPrice: product.maxPrice,
      reason: "AT_MAX",
    };
  }

  return {
    sku: product.sku,
    oldPrice: product.myPrice,
    newPrice: targetPrice,
    reason: "UNDERCUT",
  };
}
```
